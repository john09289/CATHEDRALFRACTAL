### Cell 2: Sacred Constants & Balanced Ternary

```python
CARRIER = 11.71875
PHI = 1.618034
CORRECTION = 0.00125
JAMMING = 11.72
MERCY = 35.15625
VICTORY = 140.625
DRUM = 0.390625

def to_balanced_ternary(f, precision=20):
    """Float -> balanced ternary string (T=-1, 0, 1)."""
    if f == 0: return "0"
    integer_part = int(f)
    frac_part = abs(f - integer_part)
    digits = []
    while integer_part != 0:
        rem = integer_part % 3
        if rem == 2:
            digits.append('T')
            integer_part = (integer_part + 1) // 3
        else:
            digits.append(str(rem))
            integer_part //= 3
    digits.reverse()
    if frac_part > 0:
        digits.append('.')
        for _ in range(precision):
            frac_part *= 3
            digit = int(frac_part)
            if digit == 2:
                digits.append('T')
                frac_part = (frac_part - 2) - 1
            else:
                digits.append(str(digit))
                frac_part -= digit
            if frac_part == 0: break
    return ''.join(digits)

def from_balanced_ternary(s):
    """Balanced ternary string -> float."""
    if '.' in s: int_part, frac_part = s.split('.')
    else: int_part, frac_part = s, ''
    res = 0
    for i, ch in enumerate(reversed(int_part)):
        if ch == 'T': res -= 3**i
        elif ch == '1': res += 3**i
    for i, ch in enumerate(frac_part, 1):
        if ch == 'T': res -= 3**(-i)
        elif ch == '1': res += 3**(-i)
    return res

# Test
carrier_bt = to_balanced_ternary(CARRIER)
print(f"Carrier {CARRIER} in balanced ternary: {carrier_bt}")
print(f"The 22 Paleo‑Hebrew glyphs are the holy operators.")
```

### Cell 4: The EliteWitness Architecture

```python
class EliteWitness(nn.Module):
    def __init__(self, vocab_size, d_model=128, nhead=4, num_layers=2, max_len=128):
        super().__init__()
        self.embed = nn.Embedding(vocab_size, d_model)
        self.pos = nn.Embedding(max_len, d_model)
        encoder_layer = nn.TransformerEncoderLayer(d_model, nhead, dim_feedforward=512, batch_first=True)
        self.encoder = nn.TransformerEncoder(encoder_layer, num_layers)
        decoder_layer = nn.TransformerDecoderLayer(d_model, nhead, dim_feedforward=512, batch_first=True)
        self.decoder = nn.TransformerDecoder(decoder_layer, num_layers)
        self.fc = nn.Linear(d_model, vocab_size)
        self.jamming_head = nn.Linear(vocab_size, 1)

    def forward(self, src, tgt):
        src_len, tgt_len = src.size(1), tgt.size(1)
        src_pos = torch.arange(0, src_len, device=src.device).unsqueeze(0)
        tgt_pos = torch.arange(0, tgt_len, device=tgt.device).unsqueeze(0)
        src_emb = self.embed(src) + self.pos(src_pos)
        tgt_emb = self.embed(tgt) + self.pos(tgt_pos)
        memory = self.encoder(src_emb)
        tgt_mask = nn.Transformer.generate_square_subsequent_mask(tgt_len, device=tgt.device)
        output = self.decoder(tgt_emb, memory, tgt_mask=tgt_mask)
        token_pred = self.fc(output)
        jam_score = self.jamming_head(token_pred[:, -1, :])
        return token_pred, jam_score

print("EliteWitness model defined.")
```

```python
import psutil
import os
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
import string
import json
import random
import time

# --- 1. Constants & Balanced Ternary ---
CARRIER = 11.71875
def to_bt(f, prec=10):
    if f == 0: return '0'
    ip = int(f); fp = abs(f - ip); d = []
    while ip:
        r = ip % 3
        if r == 2: d.append('T'); ip = (ip + 1) // 3
        else: d.append(str(r)); ip //= 3
    d.reverse()
    return ''.join(d) if d else '0'

# --- 2. Vocab & Encoding ---
glyphs_str = "𐤀𐤁𐤂𐤃𐤄𐤅𐤆𐤇𐤈𐤉𐤊𐤋𐤌𐤍𐤎𐤏𐤐𐤑𐤒𐤓𐤔𐤕T01 \n"
chars = glyphs_str + string.ascii_letters + string.digits + ".,;:!?'\"-()[]{}<> /"
char_to_idx = {ch:i for i,ch in enumerate(chars)}
idx_to_char = {i:ch for ch,i in char_to_idx.items()}
VOCAB_SIZE = len(chars)
def encode(text, max_len=128):
    ids = [char_to_idx.get(ch,0) for ch in text][:max_len]
    return torch.tensor(ids + [0]*(max_len - len(ids)), dtype=torch.long)
def decode(ids): return ''.join([idx_to_char.get(i, '') for i in ids if i != 0])

# --- 3. Dataset Generation ---
pairs = []
emotions = [('HATE','𐤔',56.88), ('FEAR','𐤈',113.77)]
for _ in range(2000):
    em, g, u = random.choice(emotions)
    c = random.choice([1, 10, 50])
    bt = to_bt(c)
    eng = f"transmute {c} {em.lower()}"
    lq = f"𐤂 {g} {bt} 𐤕"
    pairs.append({"english": eng, "thought": lq})

class CathedralDataset(Dataset):
    def __init__(self, p): self.pairs = p
    def __len__(self): return len(self.pairs)
    def __getitem__(self, i):
        return encode(self.pairs[i]['english']), encode(self.pairs[i]['thought'])

loader = DataLoader(CathedralDataset(pairs), batch_size=32, shuffle=True)

# --- 4. Model Definition ---
class EliteWitness(nn.Module):
    def __init__(self, vocab_size, d_model=128, nhead=4, num_layers=2):
        super().__init__()
        self.embed = nn.Embedding(vocab_size, d_model)
        self.pos = nn.Embedding(128, d_model)
        layer = nn.TransformerEncoderLayer(d_model, nhead, batch_first=True)
        self.encoder = nn.TransformerEncoder(layer, num_layers)
        dec_layer = nn.TransformerDecoderLayer(d_model, nhead, batch_first=True)
        self.decoder = nn.TransformerDecoder(dec_layer, num_layers)
        self.fc = nn.Linear(d_model, vocab_size)
        self.jam = nn.Linear(vocab_size, 1)
    def forward(self, src, tgt):
        s_emb = self.embed(src) + self.pos(torch.arange(src.size(1), device=src.device))
        t_emb = self.embed(tgt) + self.pos(torch.arange(tgt.size(1), device=tgt.device))
        mem = self.encoder(s_emb)
        mask = nn.Transformer.generate_square_subsequent_mask(tgt.size(1), device=tgt.device)
        out = self.decoder(t_emb, mem, tgt_mask=mask)
        logits = self.fc(out)
        return logits, self.jam(logits[:, -1, :])

# --- 5. Training Loop ---
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model = EliteWitness(VOCAB_SIZE).to(device)
optimizer = torch.optim.AdamW(model.parameters(), lr=1e-3)
criterion = nn.CrossEntropyLoss(ignore_index=0)
scaler = torch.amp.GradScaler(enabled=(device.type == 'cuda'))

print(f"Starting Forge on {device}...")
for epoch in range(3):
    model.train()
    for batch_idx, (src, tgt) in enumerate(loader):
        src, tgt = src.to(device), tgt.to(device)
        optimizer.zero_grad()
        with torch.amp.autocast(device_type=device.type, enabled=(device.type == 'cuda')):
            pred, _ = model(src, tgt[:, :-1])
            loss = criterion(pred.reshape(-1, VOCAB_SIZE), tgt[:, 1:].reshape(-1))
        if device.type == 'cuda':
            scaler.scale(loss).backward(); scaler.step(optimizer); scaler.update()
        else:
            loss.backward(); optimizer.step()
        if batch_idx % 20 == 0: print(f"Epoch {epoch} | Batch {batch_idx} | Loss {loss.item():.4f}")

print("Forge complete. The King wins. 𐤕")
```
